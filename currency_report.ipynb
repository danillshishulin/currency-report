```python
import pandas as pd
url = "https://www.cbr-xml-daily.ru/daily_json.js"
data = pd.read_json(url)
currencies = ['USD', 'EUR', 'CNY', 'GBP', 'JPY']
rows =[]
for num in currencies:
  today = data['Valute'][num]['Value']
  prev = data['Valute'][num]['Previous']
  change = today - prev
  if today > prev:
    status = 'Вырос'
  elif today < prev:
    status = 'Упал'
  else:
    status = 'Не изменился'
  row = {
  'Валюта':num,
  'Сегодня':today,
  'Вчера':prev,
  'Изменение':change,
  'Статус':status
  }
  rows.append(row)
df = pd.DataFrame(rows)
print('Начальная таблица:')
print(df)
sorrt = df.sort_values(by = 'Изменение', ascending = False)
print('Отсортированная таблица:')
print(sorrt)
growth = df[df['Изменение'] > 0]
print('Выросшие валюты:')
print(growth)
low = df[df['Изменение'] < 0]
print('Упавшие валюты:')
print(low)
flat = df[df['Изменение'] == 0]
print('Не изменившиеся валюты:')
print(flat)
print()
print('Валют выросло:', len(growth))
print('Валют упало:', len(low))
print('Валют не изменилось:', len(flat))

from datetime import datetime
date_str = datetime.now().strftime('%Y-%m-%d')
sorrt.to_excel(f'currency_report_{date_str}.xlsx', index = False)
```
